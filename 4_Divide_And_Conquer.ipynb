**Do you remember insertion sort?**

```python
import random

def insertion_sort_with_counter(arr):
    """
    Sorts an array using Insertion Sort and returns the sorted array
    along with the exact number of key comparisons executed.
    """
    A = arr.copy()
    n = len(A)
    comparisons = 0

    for i in range(1, n):
        key = A[i]
        j = i - 1

        # We expand the standard 'while' loop to count the evaluation
        # accurately before the break condition occurs.
        while j >= 0:
            comparisons += 1  # Counting the comparison: A[j] > key
            if A[j] > key:
                A[j + 1] = A[j]
                j -= 1
            else:
                break

        A[j + 1] = key

    return A, comparisons
```

```python
N = 20

# 1. Best Case: Array is already sorted in ascending order
best_case = list(range(1, N + 1))
print("Best Case:" + str(best_case))

# 2. Worst Case: Array is sorted in strictly descending order
worst_case = list(range(N, 0, -1))
print("Worst Case:" + str(worst_case))

# 3. Average Case: Array elements are in random order
average_case = list(range(1, N + 1))
random.shuffle(average_case)
print("Average Case:" + str(average_case))
```

```python
_, best_count = insertion_sort_with_counter(best_case)
_, worst_count = insertion_sort_with_counter(worst_case)
_, avg_count = insertion_sort_with_counter(average_case)


print(f"{'Case Scenario':<15} | {'Initial Array Status':<22} | {'Executions (N=20)':<18} | {'Theoretical Limit'}")
print("-" * 82)
print(f"{'Best Case':<15} | {'Already Sorted':<22} | {best_count:<18} | O(N)")
print(f"{'Average Case':<15} | {'Randomly Shuffled':<22} | {avg_count:<18} | O(N^2)")
print(f"{'Worst Case':<15} | {'Reverse Sorted':<22} | {worst_count:<18} | O(N^2)")

# Demonstrate the exact formulas for worst and best case executions
print("\nMathematical Verification:")
print(f"- Best Case expects N-1 comparisons: {N} - 1 = {N-1}")
print(f"- Worst Case expects N(N-1)/2 comparisons: {N} * {N-1} / 2 = {int(N*(N-1)/2)}")
```

```python
import random
import math

def merge_sort_with_counter(arr):
    """
    Sorts an array using Merge Sort and returns the sorted array
    along with the exact number of key comparisons executed.
    """
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2

    # Recursively sort both halves and accumulate their comparison counts
    left_sorted, left_count = merge_sort_with_counter(arr[:mid])
    right_sorted, right_count = merge_sort_with_counter(arr[mid:])

    # Merge the sorted halves
    merged, merge_count = merge(left_sorted, right_sorted)

    total_comparisons = left_count + right_count + merge_count
    return merged, total_comparisons

def merge(left, right):
    """
    Two-pointer merging phase. Counts exact pairwise comparisons.
    """
    result = []
    i = j = comparisons = 0

    while i < len(left) and j < len(right):
        comparisons += 1  # Counting the evaluation: left[i] <= right[j]
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Append any remaining elements (no comparisons needed for these)
    result.extend(left[i:])
    result.extend(right[j:])

    return result, comparisons
```

```python
def generate_worst_case(arr):
    """
    Recursively un-merges a sorted array to force maximum interleaving.
    This guarantees the two-pointer merge phase exhausts all possible comparisons.
    """
    if len(arr) <= 1:
        return arr
    # Separate elements into even and odd indices
    left = generate_worst_case(arr[0::2])
    right = generate_worst_case(arr[1::2])
    return left + right

N = 16

# 1. Best Case: Array is already sorted
best_case = list(range(1, N + 1))
print("Best Case:" + str(best_case))

# 2. Worst Case: Specially constructed interleaved array
# First we need a sorted array, then we un-merge it
worst_case = generate_worst_case(list(range(1, N + 1)))
print("Worst Case:" + str(worst_case))

# 3. Average Case: Randomly shuffled array
average_case = list(range(1, N + 1))
random.shuffle(average_case)
print("Average Case:" + str(average_case))
```

```python
_, best_count = merge_sort_with_counter(best_case)
_, worst_count = merge_sort_with_counter(worst_case)
_, avg_count = merge_sort_with_counter(average_case)

# ==========================================
# Output Formatting
# ==========================================
print(f"{'Case Scenario':<15} | {'Initial Array Status':<22} | {'Executions (N=16)':<18} | {'Time Complexity'}")
print("-" * 75)
print(f"{'Best Case':<15} | {'Already Sorted':<22} | {best_count:<18} | O(N log N)")
print(f"{'Average Case':<15} | {'Randomly Shuffled':<22} | {avg_count:<18} | O(N log N)")
print(f"{'Worst Case':<15} | {'Maximum Interleaved':<22} | {worst_count:<18} | O(N log N)")

# Demonstrate the exact formulas for worst and best case executions (valid for powers of 2)
print("\nMathematical Verification (For N=2^k):")
print(f"- Best Case expects (N/2) * log2(N) comparisons: ({N}/2) * {int(math.log2(N))} = {int((N/2) * math.log2(N))}")
print(f"- Worst Case expects N * log2(N) - N + 1 comparisons: {N} * {int(math.log2(N))} - {N} + 1 = {int(N * math.log2(N) - N + 1)}")
```
